```python
# Linear layer with sigmoid
# (example from lecture in code)

def sigmoid(x):
  return 1 / (1 + math.exp(-x))

# Forward pass
def forward(x0, w0, x1, w1, w2):
    s0 = w0 * x0
    s1 = w1 * x1
    s2 = s0 + s1
    s3 = s2 + w2
    L = sigmoid(s3)
    return L

# Backward pass
def backward(x0, w0, x1, w1, w2, L):
    dL = 1.0
    ds3 = dL * L * (1 - L)
    dw2 = ds3
    ds2 = ds3
    ds1 = ds2
    ds0 = ds2
    dw1 = ds1 * x1
    dx1 = ds1 * w1
    dw0 = ds0 * x0
    dx0 = ds0 * w0
    return  dx0, dw0, dx1, dw1, dw2


# Verify with example from the lecture
x0 = -1.0
w0 = 2.0
x1 = -2.0
w1 = -3.0
w2 = -3.0

L = forward(x0, w0, x1, w1, w2)
print("Output:", L)

dx0, dw0, dx1, dw1, dw2 = backward(x0, w0, x1, w1, w2, L)

print("Gradients:")
print("dx0:", dx0)
print("dw0:", dw0)
print("dx1:", dx1)
print("dw1:", dw1)
print("dw2:", dw2)
```

```python
# Two-layer NN with ReLU activation (Q2.1)

# ReLU activation function
def relu(z):
    return np.maximum(0, z)

# Forward pass
def forward(x, W1, W2, b1, b2):
    s0 = W1.T @ x
    s1 = s0 + b1
    s2 = relu(s1)
    s3 = (W2.T @ s2).item() # extracts the single value from array
    pred = s3 + b2    
    return pred, s1, s2

# Backward pass
def backward(x, W1, W2, s1, s2, pred, y):
    dL = 2.0 * (pred - y)   # some upstream scalar to start from

    # L = s3 + b2
    ds3 = dL * 1.0
    db2 = dL * 1.0

    # s3 = W2.T @ s2
    dW2 = ds3 * s2.reshape(-1, 1)
    ds2 = ds3 * W2.flatten()

    # s2 = relu(s1)
    ds1 = ds2 * (s1 > 0)

    # s1 = s0 + b1
    db1 = ds1 * 1.0
    ds0 = ds1 * 1.0

    # s0 = W1.T @ x
    dW1 = np.outer(x, ds0)
    dx = W1 @ ds0
    return dW1, db1, dW2, db2


# Loss function (Q2.2)
def MSE_loss(pred, gt):
    # TODO: Implement the mean squared error
    loss = np.mean((pred - gt)**2)
    return loss
```
